```python
app_id = "ru.subprogram.guitarsongs"
```

```python
from google_play_scraper import reviews_all, Sort
from datetime import datetime, date
import json

reviews = reviews_all(
    app_id,
    sleep_milliseconds=0,
    lang="ru",
    country="ru",
    sort=Sort.NEWEST,
)

def make_json_serializable(obj):
    if isinstance(obj, (datetime, date)):
        return obj.isoformat()
    elif isinstance(obj, dict):
        return {k: make_json_serializable(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [make_json_serializable(x) for x in obj]
    elif isinstance(obj, tuple):
        return [make_json_serializable(x) for x in obj]
    else:
        return obj

reviews_clean = make_json_serializable(reviews)

with open("guitarsongs_reviews_ru.json", "w", encoding="utf-8") as f:
    json.dump(reviews_clean, f, ensure_ascii=False, indent=2)

print(f"Saved {len(reviews_clean)} reviews")
```
